## Rock, Paper, Scissors Game

```python
print("Welcome to Rock, Paper, Scissors Game")
print("""Instructions:
1.Enter one of the choices: rock, paper, scissors.
2.The computer will also choose randomly.
3.Rules of the Game:
    -Rock beats Scissors.
    -Scissors beats Paper.
    -Paper beats Rock.
4.If both choose the same option, it's a tie.
5.After each round you will be asked if you want to play agian.
6.If you want to play again type 'yes' or 'no' to exit.
Let's start the game......""")


import random
print(" Welccome to Rock, Paper, Scissors Game")
choices = ["rock", "paper", "scissors"]
while True:
    try:
        user_input=input("Enter Rock,Paper,Scissors:").lower()
        if user_input not in choices:
            print("invalid! enter Rock, Paper, Scissors")
            continue
        computer=random.choice(choices)
        print("you chose:",user_input)
        print("computer chose:",computer)
        if user_input == computer:
            print("its a tie!")
            print("you chose:",user_input)
            print("computer chose:",computer)
        elif (
            (user_input== "rock" and computer == "scissors")or
            (user_input== "paper" and computer == "rock")or
            (user_input== "scissors" and computer == "paper")
        ):
            print('you win')
        else:
            print('computer wins')
        play_again= input('Do you want to play again? (yes/no)').lower()
        if play_again != "yes":
            print ("thanks for playing")
            break
    except Exception as e:
        print("An error occured:",)
```
